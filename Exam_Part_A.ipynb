```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import numpy as np
import matplotlib.pyplot as plt

# Integrators

def move_Euler(f, y, dt):
    return y + dt * f(y)

def move_RK2(f, y, dt):
    k1 = f(y)
    y_half = y + 0.5 * dt * k1
    k2 = f(y_half)
    return y + dt * k2

# 1-planet around fixed star at origin (GM = 1)
# State: y = [x, vx, y, vy]

def diffeq_1body(y):
    x, vx, yy, vy = y
    r = np.hypot(x, yy)
    ax = -x / r**3
    ay = -yy / r**3
    return np.array([vx, ax, vy, ay], dtype=float)

def energy_1body(y):
    x, vx, yy, vy = y
    r = np.hypot(x, yy)
    K = 0.5 * (vx*vx + vy*vy)
    U = -1.0 / r
    return K + U

def angmom_1body(y):
    x, vx, yy, vy = y
    return x * vy - yy * vx

def simulate(method, f, y0, t_range, dt):
    ys = np.zeros((len(t_range), len(y0)))
    y = y0.copy().astype(float)
    for i, t in enumerate(t_range):
        ys[i] = y
        y = method(f, y, dt)
    return ys

# Part 1 — Circular orbit, RK2, dt=0.01, run 2 periods (0..4π)

dt = 0.01
r0 = 1.0
vc = 1.0 / np.sqrt(r0)
y0 = np.array([r0, 0.0, 0.0, vc], dtype=float)

t_end = 4*np.pi
t_range = np.arange(0.0, t_end + dt, dt)

ys_rk2 = simulate(move_RK2, diffeq_1body, y0, t_range, dt)
xs, ys = ys_rk2[:,0], ys_rk2[:,2]

fig, axes = plt.subplots(1, 2, figsize=(11, 4))

# Trajectory (xy)
axes[0].plot(xs, ys)
axes[0].plot([xs[0]], [ys[0]], 'o', label='start')
axes[0].plot([0], [0], 'ko', markersize=4)
axes[0].set_aspect('equal', 'box')
axes[0].set_xlabel('x'); axes[0].set_ylabel('y')
axes[0].set_title('Part 1: RK2 circular orbit (dt=0.01)')
axes[0].legend()

# x(t), y(t)
axes[1].plot(t_range, xs, label='x(t)')
axes[1].plot(t_range, ys, label='y(t)')
axes[1].set_xlabel('t'); axes[1].set_ylabel('position')
axes[1].set_title('Part 1: x(t), y(t)')
axes[1].legend()

plt.tight_layout()
plt.show()

# Part 2 — Euler vs RK2, dt=0.2, run 5 periods

dt = 0.2
T = 2*np.pi
t_end = 5*T
t_range = np.arange(0.0, t_end + dt, dt)

y0 = np.array([1.0, 0.0, 0.0, 1.0], dtype=float)

ys_e = simulate(move_Euler, diffeq_1body, y0, t_range, dt)
ys_r = simulate(move_RK2,   diffeq_1body, y0, t_range, dt)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(ys_e[:,0], ys_e[:,2], label='Euler')
ax.plot(ys_r[:,0], ys_r[:,2], label='RK2')
ax.plot([ys_e[0,0]], [ys_e[0,2]], 'o', label='start')
ax.plot([0], [0], 'ko', markersize=4)
ax.set_aspect('equal', 'box')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.set_title('Part 2: Euler vs RK2 (dt=0.2), 5 periods')
ax.legend()
plt.show()

# Part 3 — E(t), L(t) for both methods, dt=0.2, 5 periods
# plus fractional end errors (Efinal-E0)/E0 and same for L

E0 = energy_1body(y0)
L0 = angmom_1body(y0)

E_e = np.array([energy_1body(s) for s in ys_e])
E_r = np.array([energy_1body(s) for s in ys_r])
L_e = np.array([angmom_1body(s) for s in ys_e])
L_r = np.array([angmom_1body(s) for s in ys_r])

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(t_range, E_e, label='Euler E(t)')
ax.plot(t_range, E_r, label='RK2 E(t)')
ax.axhline(E0, linestyle='--', label='E0')
ax.set_xlabel('t'); ax.set_ylabel('Energy E')
ax.set_title('Part 3: Energy vs time (dt=0.2)')
ax.legend()
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(t_range, L_e, label='Euler L(t)')
ax.plot(t_range, L_r, label='RK2 L(t)')
ax.axhline(L0, linestyle='--', label='L0')
ax.set_xlabel('t'); ax.set_ylabel('Angular momentum L')
ax.set_title('Part 3: Angular momentum vs time (dt=0.2)')
ax.legend()
plt.tight_layout()
plt.show()

frac_E_err_e = (E_e[-1] - E0) / E0
frac_E_err_r = (E_r[-1] - E0) / E0
frac_L_err_e = (L_e[-1] - L0) / L0
frac_L_err_r = (L_r[-1] - L0) / L0

print("Part 3 end fractional errors:")
print(f"  Euler: (Efinal-E0)/E0 = {frac_E_err_e:+.3e}")
print(f"  RK2:   (Efinal-E0)/E0 = {frac_E_err_r:+.3e}")
print(f"  Euler: (Lfinal-L0)/L0 = {frac_L_err_e:+.3e}")
print(f"  RK2:   (Lfinal-L0)/L0 = {frac_L_err_r:+.3e}")

# Part 4 — Kepler's 3rd law: measure T at r=0.5,1,2,4
# Measure period when planet returns to starting x-position (x=r0)
# with positive vy. Practical detection:
#   detect an upward crossing of y through 0 (y from negative to >=0)
#   AND x > 0 (right side) AND vy > 0.
# That corresponds to a full revolution for this start.

def measure_period_rk2(r0):
    vc = 1.0 / np.sqrt(r0)
    y0 = np.array([r0, 0.0, 0.0, vc], dtype=float)
    T_theory = 2*np.pi * r0**(3/2)

    dt = T_theory / 800.0  # ~800 steps/orbit (a bit tighter helps period detection)
    t_range = np.arange(0.0, 2*T_theory + dt, dt)

    y = y0.copy()
    y_prev = y.copy()

    # to avoid counting the initial condition at t=0
    for i, t in enumerate(t_range[1:], start=1):
        y_prev = y.copy()
        y = move_RK2(diffeq_1body, y, dt)

        # upward crossing of y=0:
        y_old, y_new = y_prev[2], y[2]
        if (y_old < 0.0) and (y_new >= 0.0):
            x_new = y[0]
            vy_new = y[3]
            if (x_new > 0.0) and (vy_new > 0.0):
                # linear interpolation to refine crossing time (optional but helps)
                alpha = (0.0 - y_old) / (y_new - y_old) if (y_new != y_old) else 0.0
                t_cross = (t - dt) + alpha * dt
                return t_cross, T_theory
    return None, T_theory

radii = [0.5, 1.0, 2.0, 4.0]
measured_T = []

for r0 in radii:
    Tm, Tth = measure_period_rk2(r0)
    measured_T.append(Tm)
    print(f"r={r0:>3}: T_measured={Tm:.6f}   T_theory={Tth:.6f}")

r_arr = np.array(radii, dtype=float)
T_arr = np.array(measured_T, dtype=float)
r_fine = np.linspace(0.4, 5.0, 200)

fig, ax = plt.subplots(figsize=(5.5, 4.5))
ax.loglog(r_arr, T_arr, 'o', label='Measured')
ax.loglog(r_fine, 2*np.pi*r_fine**(3/2), '--', label='Theory $T=2\\pi r^{3/2}$')
ax.set_xlabel('r'); ax.set_ylabel('T')
ax.set_title("Part 4: Kepler's 3rd law (log-log)")
ax.legend()
plt.tight_layout()
plt.show()

slope = np.polyfit(np.log(r_arr), np.log(T_arr), 1)[0]
print(f"Part 4: measured exponent n = {slope:.4f} (expected 1.5)")

# Part 5A — Elliptical orbit (start (1,0) with vy=0.8), RK2
# Track E(t), L(t) and check periapsis precession

dt = 0.01
t_end = 30 * 2*np.pi   # many orbits
t_range = np.arange(0.0, t_end + dt, dt)

y0_ell = np.array([1.0, 0.0, 0.0, 0.8], dtype=float)
ys_ell = simulate(move_RK2, diffeq_1body, y0_ell, t_range, dt)

xE, yE = ys_ell[:,0], ys_ell[:,2]
rE = np.hypot(xE, yE)
EE = np.array([energy_1body(s) for s in ys_ell])
LE = np.array([angmom_1body(s) for s in ys_ell])

# periapsis detection: local minima of r(t)
peri_idx = np.where((rE[1:-1] < rE[:-2]) & (rE[1:-1] < rE[2:]))[0] + 1
peri_angles = np.arctan2(yE[peri_idx], xE[peri_idx])

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(xE, yE)
ax.plot([xE[0]], [yE[0]], 'o', label='start')
ax.plot([0], [0], 'ko', markersize=4)
ax.set_aspect('equal', 'box')
ax.set_title('Part 5A: Elliptical orbit (vy=0.8), RK2')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend()
plt.show()

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(t_range, EE, label='E(t)')
ax.axhline(EE[0], linestyle='--', label='E0')
ax.set_title('Part 5A: Energy vs time')
ax.set_xlabel('t'); ax.set_ylabel('E')
ax.legend()
plt.tight_layout()
plt.show()

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(t_range, LE, label='L(t)')
ax.axhline(LE[0], linestyle='--', label='L0')
ax.set_title('Part 5A: Angular momentum vs time')
ax.set_xlabel('t'); ax.set_ylabel('L')
ax.legend()
plt.tight_layout()
plt.show()

if len(peri_angles) > 5:
    # unwrap angles to see drift cleanly
    ang_unwrap = np.unwrap(peri_angles)
    drift_per_orbit = (ang_unwrap[-1] - ang_unwrap[0]) / (len(ang_unwrap)-1)
    print(f"Part 5A: periapsis samples = {len(peri_angles)}")
    print(f"Part 5A: approx periapsis angle drift per orbit = {drift_per_orbit:.3e} rad")
else:
    print("Part 5A: not enough periapsis points detected to estimate precession.")

# Part 5B — Two planets around fixed star
#   Case 1: no planet-planet gravity
#   Case 2: include planet-planet gravity (m1=m2=1, G=1 for planet-planet)
# State: [x1,vx1,y1,vy1, x2,vx2,y2,vy2]

def diffeq_2planets_star_only(Y):
    x1, vx1, y1, vy1, x2, vx2, y2, vy2 = Y
    r1 = np.hypot(x1, y1); r2 = np.hypot(x2, y2)
    ax1, ay1 = -x1/r1**3, -y1/r1**3
    ax2, ay2 = -x2/r2**3, -y2/r2**3
    return np.array([vx1, ax1, vy1, ay1, vx2, ax2, vy2, ay2], dtype=float)

def diffeq_2planets_with_pp(Y, G=1.0, m=1.0):
    # star gravity (GM=1) + planet-planet gravity (G*m)
    x1, vx1, y1, vy1, x2, vx2, y2, vy2 = Y
    r1 = np.hypot(x1, y1); r2 = np.hypot(x2, y2)

    # star accel
    ax1, ay1 = -x1/r1**3, -y1/r1**3
    ax2, ay2 = -x2/r2**3, -y2/r2**3

    # planet-planet accel
    dx = x1 - x2
    dy = y1 - y2
    d = np.hypot(dx, dy)
    # accel on 1 due to 2: -G*m*(r1-r2)/|r1-r2|^3
    ax1 += -G*m * dx / d**3
    ay1 += -G*m * dy / d**3
    # accel on 2 due to 1: +G*m*(r1-r2)/|r1-r2|^3
    ax2 += +G*m * dx / d**3
    ay2 += +G*m * dy / d**3

    return np.array([vx1, ax1, vy1, ay1, vx2, ax2, vy2, ay2], dtype=float)

# initial: opposite sides, same circular speed at r=1
r0 = 1.0
vc = 1.0
Y0 = np.array([+r0, 0.0, 0.0, +vc,   -r0, 0.0, 0.0, -vc], dtype=float)

dt = 0.01
t_end = 10 * 2*np.pi
t_range = np.arange(0.0, t_end + dt, dt)

Y_star_only = simulate(move_RK2, diffeq_2planets_star_only, Y0, t_range, dt)
Y_with_pp   = simulate(move_RK2, diffeq_2planets_with_pp,   Y0, t_range, dt)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(Y_star_only[:,0], Y_star_only[:,2], label='Planet 1 (no P-P)')
ax.plot(Y_star_only[:,4], Y_star_only[:,6], label='Planet 2 (no P-P)')
ax.plot([0], [0], 'ko', markersize=4)
ax.set_aspect('equal', 'box')
ax.set_title('Part 5B: Two planets, star gravity only')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend()
plt.show()

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(Y_with_pp[:,0], Y_with_pp[:,2], label='Planet 1 (with P-P)')
ax.plot(Y_with_pp[:,4], Y_with_pp[:,6], label='Planet 2 (with P-P)')
ax.plot([0], [0], 'ko', markersize=4)
ax.set_aspect('equal', 'box')
ax.set_title('Part 5B: Two planets, WITH planet-planet gravity')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend()
plt.show()

# Part 5C — Equal-mass binary, no fixed star
# Two bodies (m1=m2=1, G=1)
# Positions: (-1,0) and (+1,0)
# Circular binary solution: speed v = sqrt(G*m/4) = 0.5
# Period should be T = 4π
# State: [x1,vx1,y1,vy1, x2,vx2,y2,vy2]

def diffeq_binary(Y, G=1.0, m=1.0):
    x1, vx1, y1, vy1, x2, vx2, y2, vy2 = Y
    dx = x1 - x2
    dy = y1 - y2
    d = np.hypot(dx, dy)

    ax1 = -G*m * dx / d**3
    ay1 = -G*m * dy / d**3
    ax2 = +G*m * dx / d**3
    ay2 = +G*m * dy / d**3
    return np.array([vx1, ax1, vy1, ay1, vx2, ax2, vy2, ay2], dtype=float)

v_circ = 0.5
Y0_bin = np.array([-1.0, 0.0, 0.0, +v_circ,   +1.0, 0.0, 0.0, -v_circ], dtype=float)

dt = 0.01
t_end = 6 * 2*np.pi   # a few cycles (expected period 4π)
t_range = np.arange(0.0, t_end + dt, dt)

Ybin = simulate(move_RK2, diffeq_binary, Y0_bin, t_range, dt)

fig, ax = plt.subplots(figsize=(6, 6))
ax.plot(Ybin[:,0], Ybin[:,2], label='Body 1')
ax.plot(Ybin[:,4], Ybin[:,6], label='Body 2')
ax.plot([0], [0], 'ko', markersize=4)
ax.set_aspect('equal', 'box')
ax.set_title('Part 5C: Equal-mass binary (RK2)')
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.legend()
plt.show()

# measure binary period: body 1 upward crossing of y=0 on left side (x<0) with vy>0
def measure_period_binary(Ytraj, t_range):
    y = Ytraj[:,2]      # y1
    x = Ytraj[:,0]      # x1
    vy = Ytraj[:,3]     # vy1
    for i in range(2, len(t_range)):
        if (y[i-1] < 0.0) and (y[i] >= 0.0) and (x[i] < 0.0) and (vy[i] > 0.0):
            # linear interpolate to y=0
            y_old, y_new = y[i-1], y[i]
            alpha = (0.0 - y_old) / (y_new - y_old) if (y_new != y_old) else 0.0
            return (t_range[i-1] + alpha*(t_range[i]-t_range[i-1]))
    return None

T_bin_meas = measure_period_binary(Ybin, t_range)
print(f"Part 5C: suggested v_circ = {v_circ:.3f}")
print(f"Part 5C: measured period ≈ {T_bin_meas:.6f}  (expected 4π = {4*np.pi:.6f})")
```
