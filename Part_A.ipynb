```
import numpy as np
import matplotlib.pyplot as plt

def move_Euler(f, y, dt):
    return y + dt * f(y)

def move_RK2(f, y, dt):
    ydot = f(y)
    y_half = y + 0.5 * dt * ydot
    ydot_half = f(y_half)
    return y + dt * ydot_half

def sir_rhs_factory(N, beta, gamma):
    """Return RHS function f(y) for SIR with fixed beta,gamma."""
    def f(y):
        S, I, R = y
        dSdt = -beta * S * I / N
        dIdt =  beta * S * I / N - gamma * I
        dRdt =  gamma * I
        return np.array([dSdt, dIdt, dRdt], dtype=float)
    return f

def run_sir(N, beta, gamma, dt, tmax, y0, method="RK2"):
    """Run SIR and return t, S, I, R arrays."""
    t = np.arange(0, tmax + 1e-12, dt)
    f = sir_rhs_factory(N, beta, gamma)

    step = move_RK2 if method.upper() == "RK2" else move_Euler

    S = np.zeros_like(t, dtype=float)
    I = np.zeros_like(t, dtype=float)
    R = np.zeros_like(t, dtype=float)

    y = y0.astype(float).copy()
    for k, _ in enumerate(t):
        S[k], I[k], R[k] = y
        y = step(f, y, dt)

    return t, S, I, R

N = 10_000

# Part 1 — Simulating an Outbreak

beta = 0.3
gamma = 0.1
dt = 0.5
tmax = 200.0

S0, I0, R0 = N - 10, 10, 0
y0 = np.array([S0, I0, R0], dtype=float)

t1, S1, I1, R1 = run_sir(N, beta, gamma, dt, tmax, y0, method="RK2")

# Plot S, I, R
fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(t1, S1, label="Susceptible S")
ax.plot(t1, I1, label="Infectious I")
ax.plot(t1, R1, label="Recovered R")
ax.set_xlabel("Days")
ax.set_ylabel("Number of individuals")
ax.set_title(f"SIR Epidemic  (β={beta}, γ={gamma}, R₀={beta/gamma:.1f})")
ax.legend()
plt.tight_layout()
plt.show()

# Report peak and final values
peak_I = float(I1.max())
day_peak = float(t1[np.argmax(I1)])
final_S = float(S1[-1])
print("\n=== Part 1 Results ===")
print(f"Peak infections: {peak_I:.0f} on day {day_peak:.1f}")
print(f"Final susceptibles: {final_S:.0f} ({100*final_S/N:.1f}% never infected)")

print("\nPart 1 Written Response (example):")
print(
    "S(t) decreases as people become infected. I(t) rises at first because infections create new "
    "infections while many susceptibles remain, then falls once S becomes small enough that transmission "
    "cannot outpace recovery. R(t) increases monotonically as infected individuals recover and accumulate. "
    "Not everyone is infected because depletion of susceptibles reduces the infection rate until the outbreak "
    "dies out, leaving S(∞) > 0."
)

# Part 2 — The Epidemic Threshold

gamma = 0.1
betas = [0.05, 0.08, 0.10, 0.12, 0.15, 0.20, 0.30]
R0_vals = [b / gamma for b in betas]
dt2 = 0.5
tmax2 = 400.0

peak_Is = []
total_Rs = []

for b in betas:
    y0 = np.array([float(N - 10), 10.0, 0.0])
    t, S, I, R = run_sir(N, b, gamma, dt2, tmax2, y0, method="RK2")
    peak_Is.append(float(I.max()))
    total_Rs.append(float(R[-1]))

fig, axes = plt.subplots(1, 2, figsize=(12, 4))

axes[0].plot(R0_vals, peak_Is, "bo-")
axes[0].axvline(1.0, color="r", linestyle="--", label="R₀=1")
axes[0].set_xlabel("R₀ = β/γ")
axes[0].set_ylabel("Peak infections")
axes[0].legend()

axes[1].plot(R0_vals, [r / N for r in total_Rs], "gs-")
axes[1].axvline(1.0, color="r", linestyle="--", label="R₀=1")
axes[1].set_xlabel("R₀ = β/γ")
axes[1].set_ylabel("Total fraction infected (R(∞)/N)")
axes[1].legend()

plt.tight_layout()
plt.show()

print("\n=== Part 2 Results ===")
for b, R0v, pI, finR in zip(betas, R0_vals, peak_Is, total_Rs):
    print(f"β={b:0.2f}  R₀={R0v:0.2f}  peak I={pI:8.1f}  total infected frac={finR/N:0.4f}")

print("\nPart 2 Written Response (example):")
print(
    "Yes—there is a clear threshold near R₀=1. For R₀<1, infections do not grow and outbreaks remain small. "
    "For R₀>1, the epidemic takes off, producing a large peak and infecting a substantial fraction of the population. "
    "In a deterministic model this transition is sharp in the sense that sustained growth requires R₀>1."
)

# Part 3 — Conservation Check

beta = 0.3
gamma = 0.1
dt3 = 0.5
tmax3 = 200.0
t3 = np.arange(0, tmax3 + 1e-12, dt3)

y0 = np.array([float(N - 10), 10.0, 0.0])

# RK2 totals
f = sir_rhs_factory(N, beta, gamma)
totals_rk2 = []
y = y0.copy()
for _ in t3:
    totals_rk2.append(y.sum())
    y = move_RK2(f, y, dt3)

# Euler totals
totals_euler = []
y = y0.copy()
for _ in t3:
    totals_euler.append(y.sum())
    y = move_Euler(f, y, dt3)

fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(t3, totals_rk2, label="RK2")
ax.plot(t3, totals_euler, label="Euler", linestyle="--")
ax.axhline(N, color="k", linestyle=":", label="Exact N")
ax.set_xlabel("Days")
ax.set_ylabel("S + I + R")
ax.legend()
plt.tight_layout()
plt.show()

max_err_rk2 = max(abs(np.array(totals_rk2) - N)) / N
max_err_euler = max(abs(np.array(totals_euler) - N)) / N
print("\n=== Part 3 Results ===")
print(f"RK2 max fractional error:   {max_err_rk2:.2e}")
print(f"Euler max fractional error: {max_err_euler:.2e}")

print("\nPart 3 Written Response (example):")
print(
    "S+I+R is conserved up to numerical error. RK2 conserves it better than Euler because RK2 is a higher-order "
    "method that reduces truncation error by sampling the slope at the midpoint. Conservation matters because S, I, "
    "and R represent real people; if the total changes significantly, the simulation would be physically meaningless."
)

# Part 4 — Flattening the Curve

gamma = 0.1
beta0 = 0.3
t_int = 30.0
dt4 = 0.5
tmax4 = 200.0
t4 = np.arange(0, tmax4 + 1e-12, dt4)

scenarios = [0.3, 0.2, 0.12, 0.08]  # include baseline (0.3)
results_part4 = []

fig, ax = plt.subplots(figsize=(10, 5))

for beta1 in scenarios:
    y = np.array([float(N - 10), 10.0, 0.0])
    Is_scenario = []
    Rs_scenario = []

    for tt in t4:
        Is_scenario.append(y[1])
        Rs_scenario.append(y[2])

        b_now = beta0 if tt < t_int else beta1

        def f_t(yy, b=b_now):
            S, I, R = yy
            return np.array([-b * S * I / N, b * S * I / N - gamma * I, gamma * I], dtype=float)

        y = move_RK2(f_t, y, dt4)

    Is_scenario = np.array(Is_scenario)
    Rs_scenario = np.array(Rs_scenario)

    peakI = float(Is_scenario.max())
    total_infected = float(Rs_scenario[-1])  # since R(0)=0
    results_part4.append((beta1, peakI, total_infected))

    label = "No intervention" if beta1 == 0.3 else f"β₁={beta1}"
    ax.plot(t4, Is_scenario, label=label)

ax.axvline(t_int, color="k", linestyle=":", label=f"Intervention at day {t_int}")
ax.set_xlabel("Days")
ax.set_ylabel("Infectious individuals I(t)")
ax.set_title("Flattening the curve")
ax.legend()
plt.tight_layout()
plt.show()

print("\n=== Part 4 Results ===")
for beta1, peakI, total_inf in results_part4:
    print(f"β₁={beta1:0.2f}  peak I={peakI:8.1f}  total infected={total_inf:8.1f}  frac={total_inf/N:0.4f}")

print("\nPart 4 Written Response (example):")
print(
    "Lowering β after day 30 reduces the peak number of simultaneous infections (stronger reductions flatten the curve more). "
    "Intervening can reduce total infections as well, but its most reliable effect is lowering and delaying the peak. "
    "Flattening matters because hospitals have limited capacity; lowering the peak helps prevent the healthcare system "
    "from being overwhelmed even if the total number of cases is not drastically changed."
)

# Part 5 — Open-Ended Exploration

# Part 5A: Herd immunity threshold
# Theory: v* = 1 - 1/R0, with R0 = beta/gamma. Here R0=3 => v*=2/3 ~ 0.6667

beta = 0.3
gamma = 0.1
R0_basic = beta / gamma
v_theory = 1 - 1 / R0_basic

dt5 = 0.5
tmax5 = 200.0

def outbreak_grows(I_series, threshold=10.0):
    """Decide if outbreak grows beyond initial size meaningfully."""
    return I_series.max() > threshold

# search smallest v for which outbreak does not grow

v_values = np.linspace(0, 0.9, 91)  # step 0.01
v_crit = None
for v in v_values:
    R_init = v * N
    I_init = 10.0
    S_init = N - I_init - R_init
    y0 = np.array([S_init, I_init, R_init], dtype=float)
    t, S, I, R = run_sir(N, beta, gamma, dt5, tmax5, y0, method="RK2")
    if not outbreak_grows(I, threshold=I_init * 1.05):  # allow tiny numerical wiggle
        v_crit = v
        break

print("\n=== Part 5A (Herd Immunity) ===")
print(f"Theory v* = 1 - 1/R0 = {v_theory:.4f} (R0={R0_basic:.2f})")
print(f"Numerical smallest v where outbreak does not grow (grid step 0.01): v ≈ {v_crit}")

# Part 5B: SIRS reinfection (immunity waning)

beta = 0.3
gamma = 0.1
xi = 0.01
dtB = 0.5
tmaxB = 600.0  # longer to see long-term behavior

def sirs_rhs_factory(N, beta, gamma, xi):
    def f(y):
        S, I, R = y
        dSdt = -beta * S * I / N + xi * R
        dIdt =  beta * S * I / N - gamma * I
        dRdt =  gamma * I - xi * R
        return np.array([dSdt, dIdt, dRdt], dtype=float)
    return f

def run_sirs(N, beta, gamma, xi, dt, tmax, y0, method="RK2"):
    t = np.arange(0, tmax + 1e-12, dt)
    f = sirs_rhs_factory(N, beta, gamma, xi)
    step = move_RK2 if method.upper() == "RK2" else move_Euler

    S = np.zeros_like(t, dtype=float)
    I = np.zeros_like(t, dtype=float)
    R = np.zeros_like(t, dtype=float)

    y = y0.astype(float).copy()
    for k, _ in enumerate(t):
        S[k], I[k], R[k] = y
        y = step(f, y, dt)
    return t, S, I, R

y0 = np.array([float(N - 10), 10.0, 0.0])
tB, SB, IB, RB = run_sirs(N, beta, gamma, xi, dtB, tmaxB, y0, method="RK2")

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(tB, SB, label="S")
ax.plot(tB, IB, label="I")
ax.plot(tB, RB, label="R")
ax.set_xlabel("Days")
ax.set_ylabel("Individuals")
ax.set_title(f"SIRS with waning immunity (β={beta}, γ={gamma}, ξ={xi})")
ax.legend()
plt.tight_layout()
plt.show()

print("\n=== Part 5B (SIRS Reinfection) ===")
print(
    "With ξ>0, recovered individuals slowly return to susceptible, replenishing S. "
    "This often produces an endemic state where I(t) approaches a nonzero level (or shows damped oscillations), "
    "rather than dying out permanently as in SIR."
)

# Part 5C: Stochastic effects (Poisson new infections)

beta = 0.3
gamma = 0.1
dtC = 0.5
tmaxC = 200.0
tC = np.arange(0, tmaxC + 1e-12, dtC)

n_runs = 20
I0_stoch = 5
threshold_dieout = 50

all_I = []
dieouts = 0

for run in range(n_runs):
    S = N - I0_stoch
    I = I0_stoch
    R = 0
    I_series = []

    for _ in tC:
        I_series.append(I)

        # expected new infections over dt: lambda = (beta * S * I / N) * dt
        lam_inf = (beta * S * I / N) * dtC
        new_inf = np.random.poisson(lam_inf)

        # expected recoveries over dt: lam_rec = (gamma * I) * dt
        lam_rec = (gamma * I) * dtC
        new_rec = np.random.poisson(lam_rec)

        # clamp to avoid negative populations
        new_inf = min(new_inf, S)
        new_rec = min(new_rec, I)

        S -= new_inf
        I += new_inf - new_rec
        R += new_rec

    I_series = np.array(I_series, dtype=float)
    all_I.append(I_series)

    if I_series.max() <= threshold_dieout:
        dieouts += 1

fig, ax = plt.subplots(figsize=(10, 5))
for I_series in all_I:
    ax.plot(tC, I_series, alpha=0.6)
ax.axhline(threshold_dieout, linestyle="--", label=f"die-out threshold I={threshold_dieout}")
ax.set_xlabel("Days")
ax.set_ylabel("Infectious individuals I(t)")
ax.set_title(f"Stochastic SIR (Poisson infections), {n_runs} runs, I0={I0_stoch}")
ax.legend()
plt.tight_layout()
plt.show()

print("\n=== Part 5C (Stochastic) ===")
print(f"Fraction of outbreaks that die out early (never exceed I>{threshold_dieout}): {dieouts}/{n_runs} = {dieouts/n_runs:.2f}")

# Qualitative answers for Part 1-5A are printed at the end of this file, after all code and plots.

r_arr = np.array([0.5, 1.0, 2.0, 4.0])
T_arr = 2*np.pi * r_arr**(3/2)

slope = np.polyfit(np.log(r_arr), np.log(T_arr), 1)[0]
print(f"""
Part 1 Answer:

The RK2 simulation produces an orbit that is extremely close to a perfect
circle when dt = 0.01. The small timestep keeps numerical errors very small,
so the trajectory remains stable over the two simulated periods.

If the timestep were increased significantly, the orbit would begin to drift.
This occurs because numerical integration introduces small errors in the
energy of the system at each step. If the energy increases, the orbit slowly
expands outward; if it decreases, the orbit would spiral inward.


Part 2 Answer:

When dt = 0.2, the Euler method produces an orbit that quickly spirals and
becomes unstable, while the RK2 orbit remains much closer to a closed circle.

This happens because Euler's method has a larger local truncation error.
Each timestep introduces a relatively large error in the velocity and
position, and these errors accumulate over time. RK2 reduces this error
by estimating the slope at the midpoint of the timestep, which makes the
method significantly more accurate and stable.


Part 3 Answer:

In the exact gravitational two-body system, both total energy and angular
momentum should be conserved because gravity is a conservative central force.

In the simulation, Euler's method shows noticeable energy drift over time,
which causes the orbit to slowly change. The energy typically increases,
which corresponds to the orbit expanding outward.

RK2 conserves energy much more accurately, so the energy remains nearly
constant with only small fluctuations. Angular momentum should also remain
constant because the gravitational force produces no torque about the
origin, and RK2 maintains this conservation much better than Euler.


Part 4 Answer:

The measured exponent from the log-log fit is approximately {slope:.3f}.
This value is very close to the theoretical prediction of 1.5 from
Kepler's Third Law, which states that T ∝ r^(3/2).

Any small difference from the theoretical value is caused by numerical
effects such as finite timestep size and the discrete method used to
detect when the orbit completes one full revolution.


Part 5A:

When the initial velocity is reduced below the circular speed, the orbit
becomes elliptical. In theory, both energy and angular momentum should
remain constant. In the numerical simulation using RK2 they remain nearly
constant, although small numerical errors can cause slight drift. Over
many orbits this can appear as a slow rotation of the ellipse
(periapsis precession).


Part 5B:

With two planets orbiting the star but without interacting with each
other, both planets follow independent circular orbits. When the
gravitational interaction between the planets is added, the orbits
become perturbed and slightly distorted due to the additional forces.


Part 5C:

In the equal-mass binary star system, both bodies orbit their common
center of mass at the midpoint between them. With the chosen units,
a velocity of about 0.5 produces a stable circular orbit and the
orbital period is approximately 4π.
""")
```
